# Report generation with Nudge

This notebook shows how Nudge can turn business data into a structured report.

The Nudge program defines the report format before calling the model. This
ensures that the result contains a title, an executive summary, key findings,
and recommendations.

The example uses Nudge's deterministic fake provider, so no API key is needed.


```python
import os

nudge_directory = os.path.join(os.environ["LOCALAPPDATA"], "Nudge")
os.environ["PATH"] = nudge_directory + os.pathsep + os.environ["PATH"]
os.environ["NUDGE_PROVIDER"] = "fake"

print(f"Nudge folder added: {nudge_directory}")
print("Nudge provider: fake")
```

```python
%%writefile report_generation.ndg
type BusinessReport = {
    title: string,
    executive_summary: string,
    key_findings: [string],
    recommendations: [string]
}

fn generate_report(data: string, audience: string) -> BusinessReport uses LLM {
    llm"""Create a concise business report from the following data.

Data:
{data}

Audience: {audience}

Include:
- A clear report title
- An executive summary
- Key findings
- Actionable recommendations"""
    with {
        schema: BusinessReport,
        model: "anthropic:sonnet-4.6",
        budget: 0.03 USD,
        retry: 2 with repair
    }
}

fn main() -> BusinessReport uses LLM {
    let data = "Monthly sales: January $42k, February $48k, March $61k. Customer retention: 91%. Support tickets: 38, down from 52 last month."
    let audience = "The operations manager of a small SaaS company"
    generate_report(data, audience)
}
```
