# Data analysis with Nudge

This notebook shows how to use Nudge to analyze quarterly revenue data.

Nudge defines a typed output schema, checks the program, compiles it to Python,
and runs it with its deterministic fake provider. No API key is required.

The example asks Nudge to identify revenue trends, describe their business
impact, and suggest actionable recommendations.

```python
import sys
import subprocess

subprocess.run(
    [sys.executable, "-m", "pip", "install", "nudge-runtime"],
    check=True,
)
```

```python
%%writefile data_analysis.ndg
type Insight = {
    category: string,
    finding: string,
    impact: string,
    confidence: float @range(0, 1)
}

type Analysis = {
    dataset: string,
    insights: [Insight],
    summary: string,
    recommendations: [string]
}

fn analyze_data(data: string, context: string) -> Analysis uses LLM {
    llm"""Analyze this data and provide insights.

Data:
{data}

Context: {context}

Identify:
- Key patterns and trends
- Business impact
- Actionable recommendations

Rate confidence for each insight from 0 to 1."""
    with {
        schema: Analysis,
        model: "anthropic:sonnet-4.6",
        budget: 0.04 USD,
        retry: 2 with repair
    }
}

fn main() -> Analysis uses LLM {
    let data = "Q1: $100k, Q2: $150k, Q3: $120k, Q4: $200k"
    let context = "Annual revenue analysis for a SaaS startup"
    analyze_data(data, context)
}
```

```python
!nudgec check data_analysis.ndg
```

```python
import os

nudge_directory = os.path.join(os.environ["LOCALAPPDATA"], "Nudge")
os.environ["PATH"] = nudge_directory + os.pathsep + os.environ["PATH"]
os.environ["NUDGE_PROVIDER"] = "fake"

print(f"Nudge folder added: {nudge_directory}")
print("Nudge provider: fake")
```
